# Convex vs Non-convex: Why Training Can Be Bumpy

## The big picture (imagine walking downhill blindfolded)

You're trying to find the lowest point of a landscape. You're blindfolded, so all you can
do is feel which way slopes down and take a step that way. Keep going until the ground is
flat — that's your "lowest point."

How well this works depends entirely on the **shape of the landscape**.

---

## Shape 1: Convex — a single smooth bowl 🥣

Picture the inside of a salad bowl.

- There's exactly **one** lowest point (the very bottom).
- No matter where you start, walking downhill **always** takes you there.
- You literally cannot get stuck anywhere else — every path leads to the bottom.

**This is the easy case.** The linear regression we trained earlier was convex — that's why
gradient descent always found the right answer, no matter the starting weights.

**Real-world example:** rolling a marble into a bowl. Drop it anywhere — it always ends up
at the bottom center.

---

## Shape 2: Non-convex — a bumpy mountain range 🏔️

Now picture real hilly terrain: lots of valleys, small pits, flat stretches, and peaks.

- There are **many** low points — some deep, some shallow.
- Walking downhill takes you to **whatever dip is nearest to where you started** —
  which might NOT be the deepest one.
- So where you *begin* changes where you *end up*.

**This is the hard case — and it's what EVERY real neural network looks like.**

**Real-world example:** rainwater falling on a mountain. Where it pools depends on where it
landed. A drop on one slope collects in one puddle; a drop on the other slope collects
somewhere completely different. Neither might be the lowest possible point on the mountain.

## The three "traps" in a bumpy landscape

**1. Local minimum — a "good enough" dip**
A small valley that *looks* like the bottom (everything around you goes up), but a much
deeper valley exists somewhere else. You settle here and never know you could've done better.
*Picture:* a small pond partway up a mountain — flat and settled, but the sea is far below.

**2. Saddle point — flat in a tricky way**
A spot that goes downhill in some directions but uphill in others. The ground feels almost
flat, so you slow way down and can get "stuck" even though it's not really a bottom.
*Picture:* a mountain pass (the low gap between two peaks) — or the middle of a Pringle chip.
Down along one axis, up along the other.

**3. Plateau — a big flat field**
A wide, nearly-flat stretch. There's barely any downhill to follow, so progress crawls to
a near-stop.
*Picture:* walking across a huge flat parking lot looking for the drain — hard to tell which
way is even slightly downhill.

## The uncomfortable truth
Real neural networks are bumpy landscapes with **millions** of dips and flat spots. So:
- You're **not guaranteed** to find the single best set of weights.
- **Where you start matters** — that's why two training runs can give different results.
- Training is **finicky** — it may not behave the same way twice.

## But it works anyway — here's why
1. In big networks, most of the good dips are **about equally good**. You don't need *the*
   perfect one — just a good one, and there are tons of them.
2. **Mini-batch noise** (from earlier) gives little random "kicks" that can **bounce you out
   of a shallow trap** into a better one.
3. **Momentum** (the rolling ball) lets you **coast through** small bumps and flat spots
   instead of stopping in them.

So the tricks we learned earlier (mini-batch, momentum) aren't just for speed — they're also
how we *cope* with a bumpy landscape.

## Why you'll care in real life
This explains things you'll actually see when training models:
- "Why did two runs give different results?" → bumpy landscape, different starting points.
- "Why is my loss stuck and barely dropping?" → probably a plateau or saddle point.
- "Why does changing the random seed change my results?" → start-dependent, because non-convex.

Knowing this means you won't panic when training is a little unpredictable — that's **normal**
for a mountain range. It's only a bowl in the easy textbook cases.

## In one line
**Convex = one smooth bowl → always find the bottom (easy).**
**Non-convex = a bumpy mountain range → you might get stuck in a shallow dip, and where you
start matters (this is every real neural network).**
That's *why* training is finicky — and *why* mini-batch noise and momentum exist: to help
escape the bumps.
