```python
import numpy as np
import time


# Eigen decomposition of a 2X2 matrix
A = np.array([[4, 1],
              [2, 3]])

# Step 1: get eigenvalues and eigenvectors
eigenvalues, eigenvectors = np.linalg.eig(A)

# Step 2: Calculate Q, Λ, Q⁻¹
Q = eigenvectors                     
Lambda = np.diag(eigenvalues)        
Q_inv = np.linalg.inv(Q)

A_reconstructed = Q @ Lambda @ Q_inv


print("\nIs reconstructed_A and original A same ?", np.allclose(A, A_reconstructed))
 
```

```python
# Now let us extract the lamba out of the equation
# Λ = Q⁻¹ · A · Q

Lambda_extracted = Q_inv @ A @ Q

print("\n Is constrcuted lamba and original lamba the same ?", np.allclose(Lambda, Lambda_extracted))
```

```python
# The final payoff of computing A^100 efficiently


# Using eigendecomposition — A^100 = Q · Λ^100 · Q⁻¹
Lambda_to_100 = np.diag(eigenvalues ** 100)   
A_to_100_decomp = Q @ Lambda_to_100 @ Q_inv


print("\nEigendecomposition A^100:")
print(A_to_100_decomp)
```

```python
# Now compare the A^100 computation using eigen decomposition with conventional approach

A_float = A.astype(float)
A_to_100 =  np.linalg.matrix_power(A_float,100)
print("\n A_to_100 ", A_to_100)

print(np.allclose(A_to_100_decomp,A_to_100))
```

```python
# Eigen decomposition of a diagonla matrix is equal to the daiganol matrix
# A diagonal matrix is already in its eigen decomposition form

D = np.array([[5,  0],
              [0, -2]])

eigenvalues_D, eigenvectors_D = np.linalg.eig(D)

Q = eigenvectors_D                     
Lambda = np.diag(eigenvalues_D)        
Q_inv = np.linalg.inv(Q)

eigen_decomp_D = Q @ Lambda @ Q_inv

print("Are those two the same ? ", np.allclose(eigen_decomp_D,D))
```

```python
# A case of defective matrix which cannot be decomposed into its eigendecompostion form

N = np.array([[2, 1],
              [0, 2]])

eigenvalues_N, eigenvectors_N = np.linalg.eig(N)

rank_of_eigen_vector_matrix = np.linalg.matrix_rank(eigenvectors_N)
dim_of_eigen_vector_matrix = eigenvectors_N.shape[1]

print("\nRank of eigenvectors matrix:", rank_of_eigen_vector_matrix)
print("No of columns/dimenisons of eigenvector_n matrix ",  dim_of_eigen_vector_matrix)
print("Are set of vectors in eigenvector_matrix is linearly dependent ?", rank_of_eigen_vector_matrix == dim_of_eigen_vector_matrix)


# Let us try to break it into eigen decompsotion form anyway


Q_N = eigenvectors_N
Lambda_N = np.diag(eigenvalues_N)
try:
    Q_N_inv = np.linalg.inv(Q_N)
    print("\nQ⁻¹:", Q_N_inv)
except np.linalg.LinAlgError as e:
    print("Cannot invert Q_N:", e)


N_reconstructed = Q_N @ Lambda_N @ Q_N_inv
print("\nIs the reconstructed N equal to the original one ?", np.allclose(N, N_reconstructed))

# This means the matrix is non-diagonlaizable
```

```python
# Product of eigen values is equal to the dterminant of the matrix

# For all three matrices today
matrices = {
    "A": np.array([[4, 1], [2, 3]]),
    "D": np.array([[5, 0], [0, -2]]),
    "N": np.array([[2, 1], [0, 2]]),
}

for name, M in matrices.items():
    det = np.linalg.det(M)
    eigs = np.linalg.eigvals(M)
    product = np.prod(eigs)
    print(f"{name}: det = {det:.2f}, eigenvalue product = {product:.2f}")
```
