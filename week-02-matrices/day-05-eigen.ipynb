```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
A = np.array([[4, 1],
              [2, 3]])
v = np.array([1, 1])

result = A @ v
print("A @ v =", result)

# Check if result is a scalar multiple of v
# If yes, what's the scalar?
print("Ratio of components:", result / v)   # both should equal the eigenvalue

print("Since both number is 5 the eigen value is 5")
```

```python
# There is a pre defined method to identify the eigen vector and eigen value in numpy 

eigenvalues, eigenvectors = np.linalg.eig(A)

print("Eigenvalues:", eigenvalues)
print("\nEigenvectors (as COLUMNS of the matrix):")
print(eigenvectors)
```

```python
# Verfiy wheter the numpy's eigen vector and eigen values satisfies the equation.

# A @ v = λ * v

for i in range(len(eigenvalues)):
    λ = eigenvalues[i]
    v = eigenvectors[:, i]
    
    left_side  = A @ v
    right_side = λ * v
    
    print(f"λ = {λ}")
    print(f"  A @ v = {left_side}")
    print(f"  λ * v = {right_side}")
    print(f"  Equal? {np.allclose(left_side, right_side)}\n")
```

```python
# For certain matrix there is no equivalent eigen vector in numpy that is represented using the letter j

# This means the vectore and eigen value is imgainary/ does not exist

R = np.array([[0, -1],
              [1,  0]])

eigenvalues, eigenvectors = np.linalg.eig(R)

print("Eigenvalues:", eigenvalues)
print("\nEigenvectors:\n", eigenvectors)
```

```python
# Plotting eigen vector and non-eigen vectors on a plot to see how they remain on the same span 
# even after tranformation

# In the below the graph 1 and graph 2 represent eigen vector before and after tranformation 
# whereas in graph 3 we are showing a non-eigen vector 

A = np.array([[4, 1],
              [2, 3]])

# Eigenvectors (using paper representatives — clearer for plotting)
v1 = np.array([1, 1])         # λ = 5
v2 = np.array([1, -2])        # λ = 2

# A random non-eigenvector
v3 = np.array([1, 0])

# Apply the transformation to each
Av1 = A @ v1
Av2 = A @ v2
Av3 = A @ v3

fig, axes = plt.subplots(1, 3, figsize=(15, 5))

for ax, v_before, v_after, label, color in [
    (axes[0], v1, Av1, 'eigenvector v₁ (λ=5)', 'red'),
    (axes[1], v2, Av2, 'eigenvector v₂ (λ=2)', 'blue'),
    (axes[2], v3, Av3, 'non-eigenvector', 'green'),
]:
    ax.quiver(0, 0, v_before[0], v_before[1], angles='xy', scale_units='xy', scale=1,
              color=color, alpha=0.5, label='before')
    ax.quiver(0, 0, v_after[0], v_after[1], angles='xy', scale_units='xy', scale=1,
              color=color, label='after A')
    ax.axhline(0, color='black', linewidth=0.5)
    ax.axvline(0, color='black', linewidth=0.5)
    ax.set_aspect('equal')
    ax.grid(True)
    ax.legend()
    ax.set_title(label)
    ax.set_xlim(-3, 6)
    ax.set_ylim(-6, 6)

plt.tight_layout()
plt.show()
```
