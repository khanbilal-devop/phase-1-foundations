```python
import numpy as np
import matplotlib.pyplot as plt

A = np.array([[2, 1],
              [0, 3]])

v = np.array([3, 4])

# Matrix and vector multiplaition

# Two approaches to do so

print("A @ v = ", A @v)
print("np.dot(A, v) =", np.dot(A, v))
```

```python
# View 1: Traditional approach thaught during shcooling row-by-row (dot products of rows of A with v)
row1_dot = np.dot(A[0], v)
row2_dot = np.dot(A[1], v)
print("Row view:    [", row1_dot, ",", row2_dot, "]")

# View 2: column picture (linear combination of A's columns, scaled by v)
col1 = A[:, 0]      # first column of A
col2 = A[:, 1]      # second column of A
result = v[0] * col1 + v[1] * col2
print("Column view:", result)
```

```python
# Matrix - Matrix multiplication

A = np.array([[1, 2],
              [3, 4]])

B = np.array([[0, 1],
              [1, 0]])

print("A @ B:\n", A @ B)   # should be [[2, 1], [4, 3]]


# Matrix multiplication is non-communtative verify

print("A @ B:\n", A @ B)
print("\nB @ A:\n", B @ A)
print("\nEqual?", np.array_equal(A @ B, B @ A))
```

```python
# 90° counterclockwise rotation matrix
R = np.array([[0, -1],
              [1,  0]])

# Vector pointing right (east)
v = np.array([1, 0])

print("Original v:", v)
print("R @ v     :", R @ v)   # should be [0, 1] — pointing up (north)
```

```python
# Visulaizing the rotation

def plot_vectors(vectors, labels, colors, title):
    plt.figure(figsize=(6, 6))
    for vec, label, color in zip(vectors, labels, colors):
        plt.quiver(0, 0, vec[0], vec[1],
                   angles='xy', scale_units='xy', scale=1,
                   color=color, label=label)
    plt.axhline(0, color='black', linewidth=1)
    plt.axvline(0, color='black', linewidth=1)
    plt.grid(True)
    plt.gca().set_aspect('equal')
    plt.xlim(-2, 2)
    plt.ylim(-2, 2)
    plt.legend()
    plt.title(title)
    plt.show()

v_original = np.array([1, 0])
v_rotated  = R @ v_original

plot_vectors(
    [v_original, v_rotated],
    ['v (original)', 'R @ v (rotated 90°)'],
    ['red', 'blue'],
    '90° counterclockwise rotation'
)
```

```python
def rotation_matrix(theta):
    return np.array([[np.cos(theta), -np.sin(theta)],
                     [np.sin(theta),  np.cos(theta)]])

# Rotate by 45°
theta = np.pi / 4    # 45 degrees in radians
R_45 = rotation_matrix(theta)

v = np.array([1, 0])
v_rotated = R_45 @ v
print("Rotated by 45°:", v_rotated)    # ≈ [0.707, 0.707]

# Visualize
plot_vectors(
    [v, v_rotated],
    ['v', 'R(45°) @ v'],
    ['red', 'blue'],
    '45° rotation'
)
```
