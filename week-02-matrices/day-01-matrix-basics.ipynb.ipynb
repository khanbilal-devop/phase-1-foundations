```python
import numpy as np

A = np.array([[1, 2],
              [3, 4]])

B = np.array([[3, 0],
              [1, 2]])

C = np.array([[1, 2, 3],
              [4, 5, 6]])

print("A:\n", A)
print("\nB:\n", B)
print("\nC:\n", C)
```

```python
# Verify shapes

print("A.shape:", A.shape)
print("B.shape:", B.shape)
print("C.shape:", C.shape) 
```

```python
print("A + B :\n", A+B)


# Matrix A and C cannot be added as these are of different shapes
try:
    print("A + C :\n", A+C)
except ValueError as e:
    print("Error : ",e)
```

```python
print("A transpose:\n", np.transpose(A))

print("\nB transpose:\n", np.transpose(B))

print("\nC transpose:\n", np.transpose(C))
```

```python
left_side  = (A + B).T
right_side = A.T + B.T

print("(A + B).T:\n", left_side)
print("\nA.T + B.T:\n", right_side)
print("\nEqual?", np.array_equal(left_side, right_side))
```

```python
# Property 1: transposing twice gets you back to A
print("A:\n", A)
print("\n(A.T).T:\n", A.T.T)
print("Equal to A?", np.array_equal(A, A.T.T))

# Property 2: transpose of scalar multiple = scalar times transpose
left  = (3 * A).T
right = 3 * A.T
print("\n(3A).T equals 3*(A.T)?", np.array_equal(left, right))
```
