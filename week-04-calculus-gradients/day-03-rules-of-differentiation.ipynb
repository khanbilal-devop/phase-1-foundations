## What this day is about

On Day 2 we computed derivatives **from the definition** — the full limit
$f'(x) = \lim_{h \to 0} \frac{f(x+h) - f(x)}{h}$ — every single time.
Correct, but slow. Day 3 replaces that grind with a handful of **rules**
that give the derivative at a glance. The rules are theorems (already
proven), so we can trust them as universal tools.

## The rules

- **Power rule:** $\frac{d}{dx}x^n = n\,x^{n-1}$  (the workhorse)
- **Constant rule:** $\frac{d}{dx}[c] = 0$
- **Constant multiple:** $\frac{d}{dx}[c \cdot f] = c \cdot f'$
- **Sum / difference:** $\frac{d}{dx}[f \pm g] = f' \pm g'$
- **Product rule:** $\frac{d}{dx}[f \cdot g] = f'g + fg'$  (not $f'g'$)
- **Quotient rule:** $\frac{d}{dx}\left[\frac{f}{g}\right] = \frac{f'g - fg'}{g^2}$

Each rule is just the limit from Day 2, pre-solved. We derived the power
rule once from the definition to confirm this, then applied all the rules
by hand.

## What we verify in code

Two independent checks that our hand-derivatives are correct:


1. **Symbolic** (SymPy, this cell): returns the derivative as a *formula*,
   confirming the *entire expression*.

## What is SymPy?

SymPy is a Python library for **symbolic mathematics** — it works with
math *symbols*, not numbers. Where normal Python takes $x = 4$ and returns
a number, SymPy takes the letter $x$ as a real variable and manipulates
expressions the way we do on paper.

- **SymPy** answers "what *is* the derivative?" → a formula $f'(x)$.

That makes SymPy the stronger check: a wrong hand-derivative could match
the numerical check at one point by coincidence, but SymPy confirms the
whole formula, term for term. Think of it as an **answer key**.

**Honest caveat:** SymPy is *not* how real ML systems differentiate. It
does algebra like a human — great for checking work. Training uses
*automatic differentiation* (Week 2 / Phase 2), built for speed at scale.

```python
import sympy as sp

x = sp.symbols('x')          # declare x as a symbol, not a number

# define the three functions symbolically (same ones from Step 2 / Part 1)
exprs = {
    "(a) 3x^2 + 2x - 7": 3*x**2 + 2*x - 7,
    "(b) x^2 * x^3":     (x**2) * (x**3),
    "(c) x^2 / (x+1)":   x**2 / (x + 1),
}

for label, expr in exprs.items():
    derivative = sp.diff(expr, x)          # differentiate symbolically
    print(f"\n{label}")
    print(f"   f'(x) = {derivative}")
```

## Step 4 — Apply: differentiating a loss function (no code, concept only)

This step connects Day 3's rules to how a model actually learns. There's no
code here — it's the reasoning that ties the derivative rules to gradient
descent. This is the single most important use of derivatives in ML.

### The pieces

- **x** — the input you feed the model (e.g. a house's size).
- **w** — the "weight," a dial the model turns. The guess is `ŷ = w · x`.
- **ŷ (y-hat)** — the model's *guess* (the hat means "predicted").
- **y** — the *true* answer from the data.
- **error = ŷ − y** — how wrong the guess is (sign = direction, size = how far off).

### The loss

We measure "how wrong" with the **squared error**:

$$L = (\hat{y} - y)^2$$

We square it so that (a) over- and under-predictions don't cancel, and
(b) big mistakes are punished more than small ones. Substituting `ŷ = w·x`:

$$L(w) = (wx - y)^2$$

Here **x and y are fixed** (they're the data); **w is the variable** —
it's the only thing training can change. So we differentiate w.r.t. w.

### Differentiating with Day 3's rules (expand, then term-by-term)

We avoid the chain rule (not covered until Week 2) by expanding first:

$$L(w) = w^2x^2 - 2xy\,w + y^2$$

Differentiate term by term (constant-multiple + power + sum rules,
treating x and y as constants):

- $w^2x^2 \;\to\; 2x^2 w$
- $-2xy\,w \;\to\; -2xy$
- $y^2 \;\to\; 0$

$$\frac{dL}{dw} = 2x^2 w - 2xy = 2x(wx - y) = 2x(\hat{y} - y)$$

### Why this matters — the learning loop

The slope of the loss is **proportional to the error** $(\hat{y} - y)$:

- Perfect guess ($\hat{y} = y$) → error 0 → slope 0 → dial doesn't move (at the minimum).
- Guess too high → slope positive → gradient descent turns w **down**.
- Guess too low → slope negative → turns w **up**.
- Bigger error → bigger slope → bigger correction.

So the derivative tells us **which way to nudge w, and by how much**, to
lower the loss — exactly the "step opposite the slope" loop from Day 2.
This derivative, $2x(\hat{y}-y)$, is the heart of how linear regression
trains. We implement it in the capstone.

**Forward pointer:** expanding by hand works for a squared term but gets
ugly for real models with nested functions. The **chain rule** (Week 2)
handles that nesting cleanly, and we'll re-derive this same loss with it.
