## The core idea

A **limit** is the value a function heads toward as the input gets
arbitrarily close to a point — even if the function never actually
reaches it there.

What the limit cares about is *where you're going*, not whether you arrive.

### Anchor example

$$f(x) = \frac{x^2 - 1}{x - 1}$$

At exactly $x = 1$ this is $0/0$ — undefined, a **hole** in the graph.
But factoring gives $x^2 - 1 = (x-1)(x+1)$, so away from the hole
$f(x) = x + 1$. As $x \to 1$ from both sides, $f(x) \to 2$:

$$\lim_{x \to 1} \frac{x^2 - 1}{x - 1} = 2$$

## One-sided limits

- $\lim_{x \to a^-} f(x)$ — approach from the **left**
- $\lim_{x \to a^+} f(x)$ — approach from the **right**

The two-sided limit exists **only if both sides agree**:

$$\lim_{x \to a} f(x) \text{ exists} \iff \lim_{x \to a^-} f(x) = \lim_{x \to a^+} f(x)$$

If they disagree, the limit **does not exist (DNE)**.

```python
import numpy as np

def f(x):
    return (x**2 - 1) / (x - 1)

# approach x = 1 from each side, getting closer each step
left  = np.array([0.9, 0.99, 0.999, 0.9999])
right = np.array([1.1, 1.01, 1.001, 1.0001])

print("approaching from left: ", f(left))
print("approaching from right:", f(right))

# Here if we substitue x as one in the given fucntion it will give us a zero (hole) in the graph

# We try to idenitfy the value for the x where x is approaching towards it not reached at x =1
# This is the concept of limit
```

```python
# Now in the below case the left and the right side never approaches to a single value
# So in such case we say limit does not exist

def g(x):
    return np.abs(x) / x

# approach x = 0 from each side
left  = np.array([-0.1, -0.01, -0.001, -0.0001])
right = np.array([ 0.1,  0.01,  0.001,  0.0001])

print("approaching from left: ", g(left))
print("approaching from right:", g(right))
```

## Apply — where limits actually show up

Honest framing: limits are **foundational**, not a tool you'll wield
directly in ML code right now. Their payoff is what they *enable*.

**What they unlock:**

- **Derivatives (Day 2)** — a derivative is literally a limit: the slope
  between two points as the gap between them shrinks to zero. No limits,
  no derivatives.
- **Gradient descent** — every training step computes derivatives (built
  on limits) to decide which way to step. The "which way is downhill"
  arrow is a limit underneath.
- **Continuity** — loss functions need to be continuous (no sudden jumps)
  for gradient-based training to behave. "Continuous" is defined using
  limits.

**Where the one-sided / DNE idea returns:** the jump in `g(x) = |x|/x`
is exactly the kind of kink that makes a function **non-differentiable**
at a point. "Both sides must agree for the limit to exist" becomes
"both sides must agree for the slope to exist." ReLU, a core neural-net
function, has precisely this kind of corner at zero.

**Bottom line:** today is the runway, not the plane. We're learning
limits so derivatives make sense tomorrow.

<details>
<summary>Answers</summary>

**A1.** The value a function heads toward as the input gets arbitrarily
close to a point — regardless of whether the function is defined there.

**A2.** Factoring gives $\frac{(x-1)(x+1)}{x-1} = x+1$ away from the hole,
which $\to 2$ as $x \to 1$. The limit cares where the function is heading,
not its value *at* the point.

**A3.** It exists only if the left-hand and right-hand limits are equal:
$\lim_{x \to a^-} f = \lim_{x \to a^+} f$.

**A4.** Left limit $= -1$, right limit $= +1$. They disagree, so the
two-sided limit **does not exist (DNE)**.

**A5.** **False.** When a limit exists it's a single number; when it
doesn't exist, there is simply no limit value — not a range.
