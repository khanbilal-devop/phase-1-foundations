# Joint dsitribution, Marginal distribution & Covariance

Everything before this looked at **one** random variable at a time. But real life comes in
pairs: height *and* weight, study hours *and* exam scores, Stock A *and* Stock B.

The interesting question is: **do they move together?**

## Covariance: do two things move together?

**Our data — 4 students:**

| Student | Study hours | Exam score |
|---|---|---|
| Amy | 2 | 50 |
| Ben | 4 | 60 |
| Cara | 6 | 70 |
| Dan | 8 | 80 |

### The recipe (4 steps)
1. Find the **average** of each variable
2. For each person, see **how far they are from each average** (keep the + / − signs!)
3. **Multiply** those two distances together
4. **Average** all the products → that's the **covariance**

### Why multiplying works
- both **below** average → (−) × (−) = **positive**
- both **above** average → (+) × (+) = **positive**
- one above, one below → (+) × (−) = **negative**

So if the two things usually agree, the products are mostly positive → **positive covariance**.
If they usually disagree, mostly negative → **negative covariance**.
If it's random, they cancel out → **near zero**.

```python
students = ["Amy", "Ben", "Cara", "Dan"]
hours  = [2, 4, 6, 8]
scores = [50, 60, 70, 80]

# STEP 1: find the average of each
total_hours = 0
for h in hours:
    total_hours = total_hours + h
mean_hours = total_hours / len(hours)

total_scores = 0
for s in scores:
    total_scores = total_scores + s
mean_scores = total_scores / len(scores)

print("average hours :", mean_hours)
print("average scores:", mean_scores)


# STEP 2 & 3: distance from average, then multiply them
products = []

for i in range(len(students)):
    hours_distance = hours[i] - mean_hours      # above or below average hours
    score_distance = scores[i] - mean_scores    # above or below average score
    product = hours_distance * score_distance
    products.append(product)
    print(students[i], "-> hours diff:", hours_distance,
          "| score diff:", score_distance, "| product:", product)

print("\nNotice: ALL the products are positive!")
print("(everyone is either below-average on both, or above-average on both)")


# STEP 4: average the products -> covariance
total = 0
for p in products:
    total = total + p

covariance = total / len(products)
print("covariance =", covariance)
print("\nPositive -> study hours and exam scores move TOGETHER")
```

## Correlation: how STRONG is the relationship?

Covariance tells us the **direction** (positive/negative), but its size is meaningless —
it depends on the units. Measure hours in minutes instead and the number changes completely,
even though nothing real changed.

**The fix:** divide by both standard deviations. The units cancel out, and the answer always
lands between **−1 and +1**:

- **+1** = perfect positive
- **+0.6** = moderately strong positive
- **0** = no relationship
- **−1** = perfect negative

$$\text{Correlation} = \frac{\text{Covariance}}{\sigma_X \times \sigma_Y}$$

```python
# find the standard deviation of each variable
var_hours = 0
var_scores = 0
for i in range(len(students)):
    var_hours = var_hours + (hours[i] - mean_hours) ** 2
    var_scores = var_scores + (scores[i] - mean_scores) ** 2

var_hours = var_hours / len(students)
var_scores = var_scores / len(students)

std_hours = var_hours ** 0.5
std_scores = var_scores ** 0.5

print("std of hours :", round(std_hours, 4))
print("std of scores:", round(std_scores, 4))

correlation = covariance / (std_hours * std_scores)
print("\ncorrelation =", round(correlation, 4))
print("It's exactly 1.0 -- a PERFECT positive relationship,")
print("because our data was perfectly straight-line (every +2 hours = +10 points).")
```

## Joint & Marginal Distributions

A **joint distribution** gives the probability of **combinations** of two variables.

**100 people, sorted by height and weight:**

| | Light | Heavy | **total** |
|---|---|---|---|
| **Short** | 35 | 5 | **40** |
| **Tall** | 15 | 45 | **60** |
| **total** | **50** | **50** | **100** |

- The **4 inner cells** = the **joint** distribution (probability of each combination)
- The **row/column totals** = the **marginal** distributions (one variable, ignoring the other)

**Why "marginal"?** Because those totals are literally written in the **margins** of the table. 🙂

**And they connect:** conditional = joint ÷ marginal
P(heavy | tall) = 45/60 — the joint cell (45) divided by the margin (60).

```python
# store the joint table as a dictionary
joint = {}
joint[("short", "light")] = 35
joint[("short", "heavy")] = 5
joint[("tall",  "light")] = 15
joint[("tall",  "heavy")] = 45

# count total people
total_people = 0
for key in joint:
    total_people = total_people + joint[key]
print("total people:", total_people)

# the JOINT probabilities
print("\nJOINT probabilities (combinations):")
for key in joint:
    print("  P(", key[0], "and", key[1], ") =", joint[key] / total_people)

# the MARGINAL probabilities (add across the other variable)
short_total = joint[("short","light")] + joint[("short","heavy")]
tall_total  = joint[("tall","light")]  + joint[("tall","heavy")]

print("\nMARGINAL for height (ignore weight):")
print("  P(short) =", short_total / total_people)
print("  P(tall)  =", tall_total / total_people)

# CONDITIONAL = joint / marginal
print("\nCONDITIONAL (zoom into the tall world):")
print("  P(heavy | tall) =", joint[("tall","heavy")], "/", tall_total,
      "=", round(joint[("tall","heavy")] / tall_total, 4))
```
