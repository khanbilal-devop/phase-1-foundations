# Probability Basics

Probability is the language of uncertainty — how we measure "how likely" something is.
This notebook covers the ground floor: what probability *is* and the rules for
combining probabilities .

## What Probability Is

**Probability** = a number from **0 to 1** measuring how likely something is.
- 0 = never happens
- 1 = always happens
- 0.5 = happens half the time

**Three words to know:**
- **Sample space** = ALL the possible outcomes.
  - Coin flip → {Heads, Tails}
  - Dice roll → {1, 2, 3, 4, 5, 6}
- **Event** = a specific outcome, or a group of outcomes.
  - "roll a 4" (one outcome) or "roll an even number" (a group: {2,4,6})
- **Probability of an event** = the number telling us how likely that event is.
  - P(roll a 4) = 1/6

**The golden rule:** all the outcome probabilities in the sample space **add up to 1** —
because *something* has to happen (total certainty = 1).

```python
import numpy as np

# Sample space of a die
sample_space = [1, 2, 3, 4, 5, 6]
print("Sample space:", sample_space)

# Probability of one outcome
p_four = 1 / len(sample_space)
print("P(roll a 4) =", round(p_four, 4))

# All outcome probabilities sum to 1
total = sum(1/len(sample_space) for _ in sample_space)
print("Sum of all probabilities =", total)

# Simulation: flip a coin 10,000 times -> P(H) should be ~0.5
flips = np.random.choice(['H', 'T'], size=10000)
print("Simulated P(Heads) over 10,000 flips:", np.mean(flips == 'H'))
```

## Combining Probabilities (the Rules)

Three ways to combine, plus one big distinction.

### NOT (complement) — "chance it DOESN'T happen"
$$P(\text{not A}) = 1 - P(A)$$
If chance of rain is 0.3, chance of no rain is 0.7. (Whatever's left over.)

### AND — "chance BOTH happen" → **multiply**
$$P(\text{A and B}) = P(A) \times P(B)$$
Two coins both Heads = 0.5 × 0.5 = 0.25.
*AND makes things LESS likely (you need both) — and multiplying numbers below 1 gives smaller.*

### OR — "chance EITHER happens" → **add**
$$P(\text{A or B}) = P(A) + P(B)$$
Roll a 1 OR a 2 = 1/6 + 1/6 = 2/6.
**Catch:** if the two events can BOTH happen, subtract the overlap so you don't double-count:
$$P(\text{A or B}) = P(A) + P(B) - P(\text{A and B})$$
Example: King OR Heart — the King of Hearts is both, so subtract it once.

### Independent vs Dependent (the key distinction)
- **Independent** = one doesn't affect the other (two coin flips). AND rule works directly.
- **Dependent** = one CHANGES the other's odds (drawing cards without putting them back).
  Here the odds shift, so you must use the *updated* probability.

**The test:** "Does the first outcome change the odds of the second?" No → independent. Yes → dependent.

```python
# NOT (complement)
p_rain = 0.3
print("P(rain) =", p_rain, "-> P(no rain) =", 1 - p_rain)

# AND (independent): two coins both heads
print("P(heads AND heads) = 0.5 * 0.5 =", 0.5 * 0.5)

# OR (can't both happen): die shows 1 or 2
print("P(roll 1 OR 2) = 1/6 + 1/6 =", round(1/6 + 1/6, 4))

# OR (can overlap): King OR Heart in a 52-card deck
p_king, p_heart, p_overlap = 4/52, 13/52, 1/52   # King of Hearts is the overlap
print("P(King OR Heart) = 4/52 + 13/52 - 1/52 =", round(p_king + p_heart - p_overlap, 4))
```

```python
# Prove the AND rule by simulation: two coins, both heads
c1 = np.random.choice(['H','T'], size=100000)
c2 = np.random.choice(['H','T'], size=100000)
print("Simulated P(H and H):", round(np.mean((c1=='H') & (c2=='H')), 4), " (theory: 0.25)")

# Show DEPENDENT events: draw 2 cards without replacement, both Aces
trials, count = 100000, 0
deck = ['A']*4 + ['x']*48
for _ in range(trials):
    draw = np.random.choice(deck, size=2, replace=False)
    if list(draw) == ['A','A']:
        count += 1
print("\nDependent events (draw 2 cards, no replacement):")
print("  Simulated P(both Aces):", round(count/trials, 5))
print("  Theory (4/52)*(3/51) =", round((4/52)*(3/51), 5))
print("  Note the 3/51 -> after drawing one Ace, the odds CHANGED (dependent!)")
```
