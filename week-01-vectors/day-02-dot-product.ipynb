```python
import numpy as np

a = np.array([2,3])
b = np.array([1,4])
```

```python
# Dot product of vector a and b

print(np.dot(a,b))
print(np.sum(a*b))
```

```python
# Length/magnitude of vectors

print(np.linalg.norm(a))
print(np.linalg.norm(b)) 
```

```python
# Angle between a and b

cos_theta = np.dot(a,b) / (np.linalg.norm(a) * np.linalg.norm(b))
theta_radian = np.arccos(cos_theta)
theta_deg = np.degrees(theta_radian)

print(f"cos(θ) = {cos_theta:.4f}")
print(f"Angle: {theta_deg:.2f} degrees")
```

```python
# Parallel check between two vectors

c = np.array([4, 6])
print("dot product", np.dot(a, c))

cos_ac = np.dot(a, c) / (np.linalg.norm(a) * np.linalg.norm(c))
cos_ac = np.clip(cos_ac, -1.0, 1.0)   
print(f"cos(θ) = {cos_ac}")
print(f"Angle: {np.degrees(np.arccos(cos_ac)):.2f} degrees")
```

```python
# Perpendicular check between vectors

d = np.array([3, -2])
print("dot product =", np.dot(a, d))

cos_ad = np.dot(a, d) / (np.linalg.norm(a) * np.linalg.norm(d))
print(f"cos(θ) = {cos_ad}")
print(f"Angle: {np.degrees(np.arccos(cos_ad)):.2f} degrees")
```

```python
# self dot product is equal to the magnitude/length of the vector

print("dot product of a.a =", np.dot(a, a))
print("Magnitude of vector a =", np.linalg.norm(a)**2)
```

```python
def cosine_similarity(v1,v2):
    cos_theta =  np.dot(v1,v2)/ (np.linalg.norm(v1) * np.linalg.norm(v2))
    return np.clip(cos_theta,-1.0,1.0)
```

```python
# Sanity checks

a = np.array([2, 3])

print(cosine_similarity(a, np.array([4, 6])))    # parallel  → 1.0
print(cosine_similarity(a, np.array([3, -2])))   # perpendic → 0.0
print(cosine_similarity(a, np.array([-2, -3])))  # opposite  → -1.0
print(cosine_similarity(a, np.array([1, 4])))    # similar   → ~0.94
```

```python
# Words vector embeddings to semnatic similarity

apple   = np.array([1.0, 1.0, 0.0, 0.0])
banana  = np.array([1.0, 0.2, 0.0, 0.0])
car     = np.array([0.0, 0.0, 1.0, 0.3])
tesla   = np.array([0.0, 0.0, 1.0, 1.0])

print(f"apple vs banana: {cosine_similarity(apple, banana):.3f}")
print(f"apple vs car:    {cosine_similarity(apple, car):.3f}")
print(f"car vs tesla:    {cosine_similarity(car, tesla):.3f}")
print(f"apple vs tesla:  {cosine_similarity(apple, tesla):.3f}")
```

```python
# Stock price movement similarity

import yfinance as yf

tickers = ['AAPL', 'MSFT', 'JPM', 'XOM']
data = yf.download(tickers, period='60d')['Close']
returns = data.pct_change().dropna()

# Each ticker's return series is a vector in ~60-dimensional space
print(f"AAPL vs MSFT (both tech): {cosine_similarity(returns['AAPL'].values, returns['MSFT'].values):.3f}")
print(f"AAPL vs JPM  (tech vs bank): {cosine_similarity(returns['AAPL'].values, returns['JPM'].values):.3f}")
print(f"AAPL vs XOM  (tech vs energy): {cosine_similarity(returns['AAPL'].values, returns['XOM'].values):.3f}")
print(f"JPM  vs XOM  (bank vs energy): {cosine_similarity(returns['JPM'].values, returns['XOM'].values):.3f}")
```
