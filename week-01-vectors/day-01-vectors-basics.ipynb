```python
import numpy as np

a = np.array([2,3])
b = np.array([1,4])

print(a+b)
print(3*a)
print(2*a -b)
```

```python
import matplotlib.pyplot as plt

plt.quiver([0,0,0], [0,0,0], [a[0],b[0],(a+b)[0]], [a[1],b[1],(a+b)[1]],
           angles='xy', scale_units='xy', scale=1, color=['r','b','g'])
plt.xlim(-1,5)
plt.ylim(-1,8)
plt.grid()
plt.show()
```
